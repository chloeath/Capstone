```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# Create a blank canvas for the mockup.
fig, ax = plt.subplots(figsize=(10, 8))
ax.set_xlim(0, 100)
ax.set_ylim(0, 100)
ax.axis('off')

# Title
ax.text(50, 95, "UNet Inference Interface for Hyperspectral Data", fontsize=14, ha='center', weight='bold')

# Instructions
ax.text(50, 90, "Upload a .mat file containing your hyperspectral image and ground truth data.", fontsize=10, ha='center')
ax.text(50, 87, "The interface will display compute time, pixel accuracy, and a comparison plot.", fontsize=10, ha='center')

# File uploader box simulation
file_uploader_box = patches.Rectangle((20, 75), 60, 8, edgecolor='black', facecolor='lightgrey', lw=1)
ax.add_patch(file_uploader_box)
ax.text(50, 79, "[Choose a .mat file]", ha='center', va='center', fontsize=12)

# File loading spinner simulation
ax.text(50, 68, "Loading file... (spinner)", fontsize=10, ha='center', color='blue')

# File loaded message simulation
ax.text(50, 65, "File loaded successfully!", fontsize=10, ha='center', color='green')

# Model loading spinner simulation
ax.text(50, 60, "Loading model... (spinner)", fontsize=10, ha='center', color='blue')
ax.text(50, 57, "Model loaded.", fontsize=10, ha='center', color='green')

# Run Inference button simulation
button_box = patches.Rectangle((40, 50), 20, 6, edgecolor='black', facecolor='lightblue', lw=1)
ax.add_patch(button_box)
ax.text(50, 53, "Run Inference", ha='center', va='center', fontsize=12)

# Inference spinner simulation
ax.text(50, 45, "Running inference... (spinner)", fontsize=10, ha='center', color='blue')

# Results display simulation
ax.text(50, 40, "Compute Time: 0.25 sec", fontsize=10, ha='center')
ax.text(50, 37, "Pixel Accuracy: 0.95", fontsize=10, ha='center')

# Plot display simulation
plot_box = patches.Rectangle((10, 5), 80, 25, edgecolor='black', facecolor='white', lw=1)
ax.add_patch(plot_box)
ax.text(50, 20, "[Plot: Input Image | Ground Truth | Predicted Mask]", ha='center', va='center', fontsize=10)

plt.show()
```
